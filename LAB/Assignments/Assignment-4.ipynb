```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline


# Dataset
data = {
    'Hours_Studied': [2, 4, 6, 8, 10],
    'Classes_Attended': [3, 4, 5, 6, 8],
    'Final_Score': [65, 70, 75, 85, 95]
}

df = pd.DataFrame(data)


# Independent variables
X = df[['Hours_Studied', 'Classes_Attended']]

# Dependent variable
y = df['Final_Score']
```

#### **OLS**

```python
ols_model = LinearRegression()

ols_model.fit(X, y)

# New student
new_student = [[7, 5]]

ols_prediction = ols_model.predict(new_student)

print("OLS / Multiple Linear Regression")
print("--------------------------------")
print("Intercept:", ols_model.intercept_)
print("Hours coefficient:", ols_model.coef_[0])
print("Classes coefficient:", ols_model.coef_[1])
print("Predicted Final Score:", ols_prediction[0])
```

#### **Ridge**

```python
ridge_model = make_pipeline(
    StandardScaler(),
    Ridge(alpha=1.0)
)

ridge_model.fit(X, y)

ridge_prediction = ridge_model.predict(new_student)

# Get Ridge model from pipeline
ridge = ridge_model.named_steps['ridge']



print("\nRidge Regression")
print("----------------")
print("Alpha:", ridge.alpha)
print("Hours coefficient:", ridge.coef_[0])
print("Classes coefficient:", ridge.coef_[1])
print("Predicted Final Score:", ridge_prediction[0])
```

#### **Comparision**

```python
comparison = pd.DataFrame({
    'Model': [
        'OLS / Multiple Linear Regression',
        'Ridge Regression'
    ],
    'Prediction': [
        ols_prediction[0],
        ridge_prediction[0]
    ]
})

print("\nComparison")
print("----------")
print(comparison)
```
